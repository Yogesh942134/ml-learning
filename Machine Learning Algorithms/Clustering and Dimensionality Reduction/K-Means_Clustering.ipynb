```python
import pandas as pd
import numpy as np
import matplotlib.pyplot  as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('data/student_clustering.csv')
```

```python
df.head()
```

```python
plt.scatter(df['cgpa'],df['iq'])
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []
for i in range(1,11):
    km = KMeans(n_clusters=i)
    km.fit(df)
    wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,11),wcss)
```

```python
X = df.iloc[:,:].values
km = KMeans(n_clusters=4)
k_mean = km.fit_predict(X)
```

```python
k_mean
```

```python
plt.scatter(X[k_mean == 0,0],X[k_mean == 0,1],color='blue')
plt.scatter(X[k_mean == 1,0],X[k_mean == 1,1],color='green')
plt.scatter(X[k_mean == 2,0],X[k_mean == 2,1],color='yellow')
plt.scatter(X[k_mean == 3,0],X[k_mean == 3,1],color='grey')
```
